```python
# modules we'll use
import pandas as pd
import numpy as np
import seaborn as sns
import datetime

# read in our data
landslides = pd.read_csv("../datasets/landslide-events/catalog.csv")

# set seed for reproducibility
np.random.seed(0)
```

```python
# infer
landslides['date_parsed'] = pd.to_datetime(landslides['date'], format='mixed')
print(landslides['date_parsed'].head())
```

```python
day_of_month_landslides = landslides['date_parsed'].dt.day
day_of_month_landslides.head()
```

```python
day_of_month_landslides = day_of_month_landslides.dropna()
sns.histplot(day_of_month_landslides, kde=False, bins=31)
```
